```python
from pyspark.sql import SparkSession
```

```python
# 1. CONFIG MINIO **AVANT** SparkSession (CRITIQUE)
spark = SparkSession.builder \
    .appName("KafkaToMinIO") \
    .master("local[*]") \
    .config("spark.executor.memory", "2g") \
    .config("spark.hadoop.fs.s3a.access.key", "minioadmin") \
    .config("spark.hadoop.fs.s3a.secret.key", "minioadmin") \
    .config("spark.hadoop.fs.s3a.endpoint", "http://host.docker.internal:9000") \
    .config("spark.hadoop.fs.s3a.path.style.access", "true") \
    .config("spark.hadoop.fs.s3a.connection.ssl.enabled", "false") \
    .config("spark.hadoop.fs.s3a.impl", "org.apache.hadoop.fs.s3a.S3AFileSystem") \
    .config("spark.hadoop.fs.s3a.aws.credentials.provider", 
            "org.apache.hadoop.fs.s3a.SimpleAWSCredentialsProvider") \
    .getOrCreate()

print("Spark + MinIO S3A configuré !")
```

```python
# 2. TEST DataFrame
df = spark.createDataFrame([
    ("2026-04-26T21:00:00", 1, 45, "Dakar"),
    ("2026-04-26T21:05:00", 2, 23, "Dakar")
], ["timestamp", "section_id", "vehicle_count", "city"])

df.show()
```

```python
# 3. ÉCRITURE S3A
df.write.mode("overwrite").parquet("s3a://traffic-bucket/bronze/test-data/")
print("MINIO S3A ÉCRITURE TERMINÉE !")
print("Vérifie http://localhost:9001 → traffic-bucket/bronze/test-data/")
```
